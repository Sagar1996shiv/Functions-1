# Functions



1.  What is the difference between a function and a method in Python?
 - In Python, the terms function and method are closely related but have a key difference:

Function:
A function is a block of reusable code that performs a specific task.

It is defined using the def keyword.

It can exist independently, outside of any class.

Example:

python
def greet(name):
    return f"Hello, {name}!"
Method:
A method is a function that is associated with an object (usually defined inside a class).

It is also defined using def, but takes self as its first parameter (for instance methods).

It is called on an object (an instance of a class).

Example:

python
class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

g = Greeter()
print(g.greet("Alice"))  # greet is a method

2.  Explain the concept of function arguments and parameters in Python.
 - In Python, parameters and arguments are related concepts but refer to different things in a function:

🔹 Parameters:
Parameters are variables listed in the function definition.

They act as placeholders for the values that the function will receive.

Example:

python
def greet(name):  # 'name' is a parameter
    print(f"Hello, {name}")
🔹 Arguments:
Arguments are the actual values you pass to a function when calling it.

Example:

python
greet("Alice")  # "Alice" is the argument

3. What are the different ways to define and call a function in Python?
 - In Python, you can define and call functions in several flexible ways. Here's a breakdown:

Ways to Define Functions
Standard Function Definition

python
def greet(name):
    return f"Hello, {name}!"
Function with Default Parameters

python
def greet(name="Guest"):
    return f"Hello, {name}!"
Function with Variable-Length Arguments

python
def print_names(*names):
    for name in names:
        print(name)
Function with Keyword Arguments

python
def describe_person(name, age):

Function with Arbitrary Keyword Arguments

python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
Lambda (Anonymous) Function

python
square = lambda x: x * x
Ways to Call Functions
Positional Arguments

python
greet("Alice")
Keyword Arguments

python
describe_person(age=30, name="Bob")
Using Default Parameters

python
greet()  # Uses default value "Guest"
Unpacking Arguments from a List or Tuple

python
args = ("Alice",)
greet(*args)
Unpacking Keyword Arguments from a Dictionary

python
info = {"name": "Charlie", "age": 25}
describe_person(**info)

5. What are iterators in Python and how do they differ from iterables?
 - iterators and iterables are closely related, but they have distinct roles:

🔹 Iterable
An iterable is any object that can return an iterator.

It has a __iter__() method.

Examples: lists, tuples, dictionaries, sets, strings.

python
my_list = [1, 2, 3]
for item in my_list:  # my_list is iterable
    print(item)
🔹 Iterator
An iterator is an object that produces items one at a time from an iterable.

It must implement two methods:

__iter__() (returns itself)

__next__() (returns the next item or raises StopIteration when done)

python
my_list = [1, 2, 3]
iterator = iter(my_list)  # Get iterator from iterable

print(next(iterator))  # Output: 1
print(next(iterator))  # Output: 2

Key Differences
Feature	           Iterable	              Iterator
Definition	   Can return an iterator	    Produces next item on demand
Key method(s)	     __iter__()	            __iter__() and __next__()
Examples	    list, tuple, dict, string	   Object from iter()
Can use for?	       Yes	                 Yes
Reusable?	           Yes	                 No (once exhausted)

6. Explain the concept of generators in Python and how they are defined.
 - A generator is a function that uses the yield statement instead of return.

Each time yield is called, the function pauses, saving its state, and resumes from there on the next call.

Generators are iterators, so they can be used in loops and with next().

Defining a Generator Function
python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1
This function does not run immediately. Instead, it returns a generator object.

Using a Generator
python
gen = count_up_to(3)
print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
print(next(gen))  # Raises StopIteration

7. What are the advantages of using generators over regular functions?
 - Generators yield one item at a time, so they don't store the entire sequence in memory.

Ideal for working with large datasets or infinite sequences.

Example:

python
def big_range():  # Generator
    for i in range(1000000):
        yield i
Contrast with:

python
def big_list():  # Regular function
    return [i for i in range(1000000)]

8.  What is a lambda function in Python and when is it typically used?
 - A lambda function in Python is a small, anonymous (unnamed) function defined using the lambda keyword.
 Example:
python
square = lambda x: x * x
print(square(5))  # Output: 25
This is equivalent to:

python
def square(x):
    return x * x   
Lambda functions are typically used for short, throwaway functions, especially when you:

Don’t want to define a full def function.

Need a quick function for functional programming tools like:

map()

filter()

sorted()

reduce() (from functools)

9.  Explain the purpose and usage of the `map()` function in Python
 - The map() function in Python is used to apply a function to every item in an iterable (like a list, tuple, etc.)
 Purpose:
To transform or process items in a collection without using explicit loops.
Example:
python
def square(x):
    return x * x

nums = [1, 2, 3, 4]
result = map(square, nums)
print(list(result))  # Output: [1, 4, 9, 16]

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
 - The functions map(), reduce(), and filter() are built-in (or in functools) and are commonly used for functional-style programming in Python.

 🔹 map(function, iterable)
Purpose: Transforms each element in the iterable using the function.

Returns: A new iterator with transformed values.

Example:

python
nums = [1, 2, 3]
squared = map(lambda x: x * x, nums)
print(list(squared))  # Output: [1, 4, 9]
🔹 filter(function, iterable)
Purpose: Filters out items in the iterable where function(item) is False.

Returns: A new iterator with only the values for which the function returns True.

Example:

python
nums = [1, 2, 3, 4]
evens = filter(lambda x: x % 2 == 0, nums)
print(list(evens))  # Output: [2, 4]
🔹 reduce(function, iterable)
Purpose: Reduces the iterable to a single value by cumulatively applying the function.

Returns: A single result.

Note: It must be imported from functools.

Example:

python
from functools import reduce

nums = [1, 2, 3, 4]
total = reduce(lambda x, y: x + y, nums)
print(total)  # Output: 10
